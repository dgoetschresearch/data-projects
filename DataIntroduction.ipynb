```python
import pandas as pd
solubility_simple = pd.read_csv("solubility_simple.csv")
solubility_simple
```

```python
import plotly.express as px
solubility_scatter = px.scatter_matrix(solubility_simple,
    dimensions=["MolWt", "NumValenceElectrons", "MolMR", "MolLogP", "TPSA"],
    color="Solubility",
    title="Solubility Scatter Plot Matrix",
    color_continuous_scale="GnBu")
solubility_scatter.show()
```
